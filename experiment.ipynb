# 🧪 Demo: Phân loại Spam với 3 mô hình (Tiền xử lý: None)

Trong demo này, chúng ta sẽ chạy thử 3 thuật toán: **Random Forest**, **Naive Bayes**, và **SVM** trên tập dữ liệu thô (chỉ xử lý NaN).

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt

# Import các module tự xây dựng
from core.processor import apply_preprocessing
from core.tf_idf import TextVectorizer
from core.train import SpamClassifier, run_experiment

print("✅ Import thành công!")
```

## 1. Chuẩn bị dữ liệu (None Preprocessing)

```python
# Đọc dữ liệu gốc
df = pd.read_csv('data/train.csv')

# Áp dụng tiền xử lý None (sử dụng module processor mới)
df_none = apply_preprocessing(df, method='none')

# Chia tập Train/Test (80/20)
X_train_raw, X_test_raw, y_train, y_test = train_test_split(
    df_none['Comment'], 
    df_none['Label'], 
    test_size=0.2, 
    random_state=42
)

print(f"Số lượng mẫu Train: {len(X_train_raw)}")
print(f"Số lượng mẫu Test: {len(X_test_raw)}")
```

## 2. Trích xuất đặc trưng (TF-IDF)

```python
vectorizer = TextVectorizer(max_features=2500)
X_train = vectorizer.fit_transform(X_train_raw)
X_test = vectorizer.transform(X_test_raw)

print(f"Kích thước ma trận TF-IDF: {X_train.shape}")
```

## 3. Chạy Demo 3 Mô hình

```python
models = ['random_forest', 'naive_bayes', 'svm']
results = run_experiment(X_train, X_test, y_train, y_test, models_to_run=models)
```

## 4. So sánh kết quả

```python
# Chuyển kết quả về DataFrame để vẽ đồ thị
df_results = pd.DataFrame({
    'Model': results.keys(),
    'Accuracy': [r['accuracy'] for r in results.values()],
    'F1-Score': [r['f1_score'] for r in results.values()]
})

display(df_results)

# Vẽ biểu đồ so sánh
df_results.plot(x='Model', y=['Accuracy', 'F1-Score'], kind='bar', figsize=(10, 5))
plt.ylim(0.5, 1.0)
plt.title('So sánh hiệu năng 3 mô hình (Preprocessing: None)')
plt.ylabel('Score')
plt.xticks(rotation=0) # Chữ nằm ngang
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```
